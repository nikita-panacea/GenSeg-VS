```python
import pandas as pd
# csv_path= '/home/ubuntu/Documents/Nafisha/GenSeg-VS/GenSeg_VS.csv'
csv_path = '/home/ubuntu/Documents/Nafisha/GenSeg-VS/GenSeg_VS_omen.csv'
df= pd.read_csv(csv_path)
df.columns
```

```python
df = df.drop(columns=['image_path', 'SegmentationPath'])
df
```

```python
for idx, row in df.iterrows():
    image_path= row['image_nifti']
    mask_path= row['mask_nifti']
    df.loc[idx, 'image_nifti']= image_path.replace('omen', 'ubuntu')
    df.loc[idx, 'mask_nifti']= mask_path.replace('omen', 'ubuntu')
    # print(df['image_nifti'])
    # break
```
